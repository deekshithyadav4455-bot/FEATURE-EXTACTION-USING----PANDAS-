Extracting episodes

```python
def extract_episodes(txt):
    data = ""
    start = False
    for i in range(len(txt)):
        if txt[i] == "(":
            start = True
        if start ==True:
            data += txt[i]
        if txt[i] == ")":
            break
    return data

```

```python
df["Episodes"]=df['Title'].apply(extract_episodes)
df
```

# 2)making a new column for time stamps

```python
def extracting_monthperiod (text):
  start=False
  data=""
  for i in range(len(text)):
    if text[i]==")":
      start=True
      for j in range(i+1,i+20):
        data+=text[j]
      return data


```

```python
df["Time-period"]=df["Title"].apply(extracting_monthperiod)
df.head()
```

# Extracting the total months for each anime

```python
from datetime import datetime
from dateutil.relativedelta import relativedelta

def month_difference(txt):
    start_txt, end_txt = txt.split("-")

    start_txt = start_txt.strip()
    end_txt = end_txt.strip()

    start_date = datetime.strptime(start_txt, "%b %Y")
    end_date = datetime.strptime(end_txt, "%b %Y")

    diff = relativedelta(end_date, start_date)

    return diff.years * 12 + diff.months
```

```python
df["Total-months"] = df["Time-period"].apply(month_difference)
df.head()
#that above code is chatgpt and it used different librarires and gave that code
```
